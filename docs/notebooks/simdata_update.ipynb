目的：将模拟数据更新到最新

数据的来源有三个途径：
+ csv文件
+ 数据文件
+ IB
+ Barchart

初步思路：
1. 从csv获取历史数据
2. 从barchart获取上个月csv缺失的
3. 从IB获取实时数据

```python
# 以恒生指数为例做个案例：

from sysdata.sim.db_futures_sim_data import dbFuturesSimData
from datetime import datetime, timedelta

instrument_code = "HANG"
data=dbFuturesSimData()
dateColumn = data.get_raw_price(instrument_code).keys()
lastDate = dateColumn.max()
firstDate = dateColumn.min()
print("lastDate: ", lastDate)
print("firstDate: ", firstDate)
today = datetime.today()

# 从IB获取lastDate到today的数据

today - lastDate



```

```python
# plot price

import matplotlib.pyplot as plt
import matplotlib

# 标题里有中文，需要显式指定带中文字形的字体，否则会画成方框
matplotlib.rcParams["font.sans-serif"] = ["Hiragino Sans GB", "Arial Unicode MS", "PingFang SC", "Heiti TC", "DejaVu Sans"]
matplotlib.rcParams["axes.unicode_minus"] = False

raw_price = data.get_raw_price(instrument_code)

fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# 全历史
raw_price.plot(ax=axes[0], title=f"{instrument_code} 原始价格（全历史，{firstDate:%Y-%m-%d} 至 {lastDate:%Y-%m-%d}）")
axes[0].axvline(lastDate, color="red", linestyle="--", linewidth=1, label=f"sim data 最新 {lastDate:%Y-%m-%d}")
axes[0].legend()
axes[0].set_xlabel("")
axes[0].set_ylabel("price")

# 近期放大，便于核对是否推进到最新
raw_price.tail(500).plot(ax=axes[1], title=f"{instrument_code} 最近 500 个数据点")
axes[1].axvline(lastDate, color="red", linestyle="--", linewidth=1)
axes[1].set_xlabel("date")
axes[1].set_ylabel("price")

plt.tight_layout()
plt.show()
```
