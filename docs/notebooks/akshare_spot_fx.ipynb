```python
import akshare as ak
import pandas as pd
import os.path
from sysproduction.data.currency_data import fxPricesData

forex_hist_em_df = ak.forex_hist_em(symbol="HKDUSD")

# convert to fx_prices_csv format
# two columns: DATETIME(%Y-%m-%d %H:%M:%S) and PRICE
forex_hist_em_df['DATETIME'] = forex_hist_em_df['日期'].apply(lambda x: x.strftime('%Y-%m-%d %H:%M:%S'))
forex_hist_em_df['PRICE'] = forex_hist_em_df['最新价']
fx_prices_csv_df = forex_hist_em_df[['DATETIME', 'PRICE']]


def merge_fx_prices(instrument_code, new_df, dst_dir):
    dst_path = os.path.join(dst_dir, instrument_code + ".csv")
    
    if not os.path.exists(dst_path):
        # copy file        
        dst_df = new_df
    else:   
        # read csv from from_path        
        dst_df = pd.read_csv(dst_path)               
        
        # insert new_df into dst_df if datetime is greater than max(dst_df.datetime)
        #new_df['DATETIME'] = pd.to_datetime(new_df['DATETIME'])
        #dst_df['DATETIME'] = pd.to_datetime(dst_df['DATETIME'])
        new_df = new_df[new_df['DATETIME'] > dst_df['DATETIME'].max()]
        dst_df = pd.concat([dst_df, new_df], ignore_index=True)
        
        # write to dst_path
    dst_df.to_csv(dst_path, index=False)    
    
    
        
DATA_PATH = "/Users/henrik/dev/pysystemtrade/data/futures/fx_prices_csv"
merge_fx_prices("HKDUSD", fx_prices_csv_df, DATA_PATH)





```
